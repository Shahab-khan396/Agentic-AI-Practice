# Streaming 

```python
model.invoke("write the 200 words paragraph on Agentic AI ")
```

```python
for chunk in model.stream("write the 200 words paragraph on Agentic AI"):
    print(chunk.text,flush=True,end="|")
```

```python
from pydantic.v1.typing import resolve_annotations
response=model.batch([
    "why is the parrot have colorfull feature",
    "what are guardrail",
    "what is model training"

],
config={'max_concurrency':5,}
)
for response in response:
    print(response)
```
